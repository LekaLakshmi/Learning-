# Comparing Numpy and Python List


# Multiply the list of number with 2 and then repeat it for 10,000 times
* NumPy is faster than Python lists
* NumPy uses less memory than lists
* NumPy has many built-in mathematical and scientific functions
* NumPy arrays usually contain only one datatype, while Python lists can contain different datatypes
* can do direct operation instead of having for loop
  

```python
import numpy as np
import time
# List
a=[1,2,3,4,5,6,7,8,9,0]*10000000
start= time.time() # time.time() returns the current time in seconds
for i in range(len(a)):
    a[i]=a[i]*2
print("Time taken for list " + str(time.time() - start))


# numpy
a=[1,2,3,4,5,6,7,8,9,0]*10000000
b=np.array(a)
start=time.time()
b=b*2
print("time take for numpy " + str(time.time() - start))
```

# Numpy Data type and type Casting

```python
arr=np.array([1,2,3,4,56,76,78,78,787,314,432]) # creating numpy array manually
arr
```

```python
# if we change even one element datatype, every element datatype changes
import numpy as np
arr=np.array([1,2,3.4,4,56,76,78,78,787,314,432]) 
print(arr)

arr=np.array([1,'34',2,3.4,4,56,76,78,78,787,314,432]) 
print(arr)
print(type(arr))
```

```python
list=['string',0,3.5] # creating a list and then converting list to numpy array
arr=np.array(list) # here all elements are getting converted into string
arr
```
