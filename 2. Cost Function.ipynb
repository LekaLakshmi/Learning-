# Cost function for linear regression with one variable. 
## Problem Statement
Model which can predict housing prices given the size of the house.  


| Size (1000 sqft)     | Price (1000s of dollars) |
| -------------------| ------------------------ |
| 1                 | 300                      |
| 2                  | 500                      |



```python
import numpy as np
```

```python
# Split the dataset into features and labels
train_features=np.array([1.0,2.0])
train_labels=np.array([300.0,500.0])
```

## Computing Cost
Cost is a measure of how well our model is predicting the target price of the house.

The equation for cost with one variable is:
  $$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2 \tag{1}$$ 
 
where 
  $$f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{2}$$

```python
# Define cost function
def cost_MSE(w,b,x,y):
    m=x.shape[0]
    prediction=np.zeros(m)
    error=np.zeros(m)
    sum_error=0
    for i in range(m):
        prediction[i]=w*x[i]+b
        error[i]=(y[i]-prediction[i])**2
        sum_error+=error[i]
    MSE=sum_error/(2*m)
    return prediction,sum_error,MSE


# call the cost function for differnt weight and bias
w=100
b=100
prediction,sum_error,MSE=cost_MSE(w,b,train_features,train_labels)
print(f"Model predictions are {prediction} sum_error: {sum_error} and MSE is {MSE} ")


w=230
b=10
prediction,sum_error,MSE=cost_MSE(w,b,train_features,train_labels)
print(f"Model predictions are {prediction} sum_error: {sum_error} and MSE is {MSE} ")


w=200
b=100
prediction,sum_error,MSE=cost_MSE(w,b,train_features,train_labels)
print(f"Model predictions are {prediction} sum_error: {sum_error} and MSE is {MSE} ")       
```
