# Dot product

```python
# create function for dot product
def dot_product(a,b):
    m=a.shape[0]
    dot=0
    for i in range(m):
        dot+=a[i]*b[i]
    return dot

# find dot product using function
a=np.array([5,7])
b=np.array([2,3])
print(f"Dot product using my function {dot_product(a,b)}  ")

# Find dot product with build in function
print(f"Dot product using build in function {np.dot(a,b)}  ")
```

# Need for speed: vector vs for loop

* vectorization provides a large speed 

```python
a=np.random.rand(100000000)
b=np.random.rand(100000000)
tic=time.time()
result=dot_product(a,b)
toc=time.time()
print(f"Time taken by for loop:{toc-tic}  ")
tic=time.time()
result=np.dot(a,b)
toc=time.time()
print(f"Time taken by vector :{toc-tic}  ")
del(a)
del(b)
```

# Indexing

```python
a=np.arange(10).reshape(-1,5)
print(f" Matrix: \n{a} ")

# access one element
print(f"Access one element {a[1][0]}  ")

# access one row
print(f"Access one row {a[1]} ")

# access one column
print(f"Access one column {a[:,2]} ")
```

# Slicing

```python
a=np.arange(20).reshape(-1,4)
print(a)

# slice row and column - start:stop:step
print(a[:,1:3:1])
print(a[0:2:1,:])

# get all rows and column
print(a[:,:])
```
