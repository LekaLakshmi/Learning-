# Data Quality metric

# 1. Number of Samples

**Number of Samples** = Number of observations/data points (rows) in the dataset.

More samples → More information available → Potentially better estimates of the underlying patterns.

| Age | Salary | Experience |
| ---: | -----: | ---------: |
| 25 | 50,000 | 2 |
| 30 | 65,000 | 5 |
| 40 | 80,000 | 10 |
| 35 | 70,000 | 7 |

**Number of Samples = 4**

### Python

`number_of_samples = X.shape[0]`

---

# 2. Number of Features

**Number of Features** = Number of input variables used to describe each sample.

The **target/output column is not counted as an input feature**.

| Age | Salary | Experience | Target |
| ---: | -----: | ---------: | -----: |
| 25 | 50,000 | 2 | 0 |
| 30 | 65,000 | 5 | 1 |
| 40 | 80,000 | 10 | 1 |

- Age → Feature
- Salary → Feature
- Experience → Feature
- Target → Output

**Number of Features = 3**

### Python

`number_of_features = X.shape[1]`

---

# 3. Sample-to-Feature Ratio

**Sample-to-Feature Ratio** = Number of samples available for each feature.

### Formula

$$
\text{Sample-to-Feature Ratio} =
\frac{\text{Number of Samples}}{\text{Number of Features}}
$$

### Example

Suppose a dataset has:

- Number of Samples = 1000
- Number of Features = 20

$$
\text{Sample-to-Feature Ratio} = \frac{1000}{20} = 50
$$

**Sample-to-Feature Ratio = 50**

This means:

**For every 1 feature, we have 50 samples.**

### Why It Matters

Consider two datasets:

| Dataset | Samples | Features | Sample-to-Feature Ratio |
| ------- | ------: | -------: | ----------------------: |
| A | 1000 | 10 | 100 |
| B | 1000 | 500 | 2 |

- Dataset A → 100 samples for every feature.
- Dataset B → 2 samples for every feature.

A **higher ratio** means we have more observations relative to the number of features.

A **low ratio** means the dataset has many features compared with the number of samples, which can make learning more difficult and may increase the risk of overfitting.

There is **no universal value** that defines a good Sample-to-Feature Ratio. It depends on the dataset and machine learning model.

### Python

`sample_feature_ratio = X.shape[0] / X.shape[1]`

# 4. Missingness Rate

**Missingness Rate** = Percentage of values in the dataset that are missing.

Missing values can appear as `NaN`, `None`, or `NULL`.

### Formula

$$
\text{Missingness Rate} =
\frac{\text{Number of Missing Values}}
{\text{Total Number of Values}}
\times 100
$$

### Example

| Age | Salary | Experience |
| ---: | -----: | ---------: |
| 25 | 50,000 | 2 |
| 30 | Missing | 5 |
| Missing | 80,000 | 10 |
| 35 | 70,000 | 7 |

- Number of Samples = 4
- Number of Features = 3
- Total Values = $4 \times 3 = 12$
- Missing Values = 2

$$
\text{Missingness Rate}
=
\frac{2}{12} \times 100
=
16.67\%
$$

**Missingness Rate = 16.67%**

This means **16.67% of the values in the dataset are missing**.

### Why It Matters

Missing data means some information about the samples is unavailable.

A dataset with a high missingness rate may require preprocessing such as:

- Removing rows or columns
- Filling missing values (Imputation)
- Using models that can handle missing values

Therefore, missingness can be used as an **intrinsic dataset quality characteristic**.

### Overall Missingness

Overall missingness measures missing values across the entire dataset.

`missingness_rate = (data.isnull().sum().sum() / data.size) * 100`

### Per-Feature Missingness

We can also calculate the missingness rate separately for each feature.

| Feature | Missingness Rate |
| ------- | ---------------: |
| Age | 25% |
| Salary | 25% |
| Experience | 0% |

`missingness_per_feature = data.isnull().mean() * 100`




# 5. Duplicate Sample Rate

A duplicate sample is when the same observation/row appears more than once in the dataset.

| Age | Salary |
| --: | -----: |
|  25 |  50000 |
|  30 |  60000 |
|  25 |  50000 |
|  40 |  80000 |

Too many duplicate samples mean the dataset contains **repeated information rather than new information**.

For example:

- Dataset A → 10,000 different samples
- Dataset B → 10,000 samples, but 5,000 are duplicates

Both datasets have 10,000 rows, but Dataset A contains much more **unique information**.

## Formula

$$
\text{Duplicate Sample Rate}
=
\frac{\text{Number of duplicate samples}}
{\text{Total number of samples}}
\times 100
$$

A higher duplicate sample rate means that more of the dataset consists of repeated information.

# 6. Coefficient of Variation

## Mean

Mean = average value of the data.

## Standard Deviation

Standard deviation tells us **how much the values are spread around the mean**.

- Small standard deviation → data points are close to the mean.
- Large standard deviation → data points are more spread out from the mean.

Standard deviation has the **same unit as the feature**.

For example:

- Age → standard deviation is measured in years.
- Salary → standard deviation is measured in dollars.

Because different features can have different units and scales, directly comparing their standard deviations can be difficult.

# 6.Coefficient of Variation

Coefficient of Variation (CV) measures the **variability of a feature relative to its mean**.

$$
CV = \frac{\text{Standard Deviation}}{\text{Mean}}
$$

Since the units cancel out, CV is **unitless**.

This makes it useful for comparing the relative variability of different features.

- Low CV → less variability relative to the mean.
- High CV → more variability relative to the mean.

### Example

Feature A:

Mean = 100  
Standard Deviation = 10

$$
CV_A = \frac{10}{100} = 0.1
$$

Feature B:

Mean = 100  
Standard Deviation = 40

$$
CV_B = \frac{40}{100} = 0.4
$$

Feature B has higher variability relative to its mean.

# 7. Skewness

Skewness tells us about the **shape/asymmetry of the distribution of a feature**.

If we plot the values of one feature, skewness tells us whether the distribution is symmetric or stretched more toward one side.

- Symmetric → Skewness ≈ 0
- Right skew / Positive skew → Skewness > 0
- Left skew / Negative skew → Skewness < 0

### Right Skew

Example:

1, 2, 3, 4, 5, 100, 101

There are extreme values on the right side.

Therefore:

Skewness > 0

### Left Skew

Example:

1, 101, 102, 103, 104, 107, 111

There is an extreme value on the left side.

Therefore:

Skewness < 0

## Formula

$$
\text{Skewness}
=
\frac{1}{N}
\sum_{i=1}^{N}
\left(
\frac{x_i-\mu}{\sigma}
\right)^3
$$

Where:

- $x_i$ = each data point
- $\mu$ = mean
- $\sigma$ = standard deviation
- $N$ = number of data points

## Why do we cube?

### Raw difference

$$
x_i - \mu
$$

Raw differences cancel each other out.

$$
\sum (x_i-\mu)=0
$$

Therefore, raw differences cannot tell us the skewness.

### Square

$$
(x_i-\mu)^2
$$

Squaring removes the sign.

Both negative and positive differences become positive, so we cannot determine whether the distribution is left-skewed or right-skewed.

### Cube

$$
(x_i-\mu)^3
$$

Cubing preserves the sign:

- Negative difference → remains negative
- Positive difference → remains positive

Cubing also makes large deviations from the mean much more important.

Therefore:

- Large positive cubed deviations → Positive / Right skew
- Large negative cubed deviations → Negative / Left skew

# 8. Kurtosis

Kurtosis tells us how strongly a distribution contains **extreme values relative to its overall spread**.

- Higher kurtosis → heavier tails / stronger tendency for extreme values.
- Lower kurtosis → lighter tails / less tendency for extreme values.

Even a few values that are very far from the mean can greatly increase kurtosis.

## Why Power 4?

We use power 4 because:

- We do not care whether the extreme value is on the left or right side.
- Power 4 removes the sign because both positive and negative values become positive.
- Power 4 gives much more importance to values that are far from the mean.

For example:

$$
2^4 = 16
$$

$$
10^4 = 10000
$$

So values far from the mean have a very large effect on kurtosis.

## Formula

The formula is similar to skewness, but we use power 4 instead of power 3.

$$
\text{Kurtosis}
=
\frac{1}{N}
\sum_{i=1}^{N}
\left(
\frac{x_i-\mu}{\sigma}
\right)^4
$$

Where:

- $x_i$ = each data point
- $\mu$ = mean
- $\sigma$ = standard deviation
- $N$ = number of data points

### Easy Difference

- **Skewness → Power 3 → direction of asymmetry**
- **Kurtosis → Power 4 → heaviness of tails / extreme values**
