# Data Quality metric

# 1. Number of Samples

**Number of Samples** = Number of observations/data points (rows) in the dataset.

More samples → More information available → Potentially better estimates of the underlying patterns.

| Age | Salary | Experience |
| ---: | -----: | ---------: |
| 25 | 50,000 | 2 |
| 30 | 65,000 | 5 |
| 40 | 80,000 | 10 |
| 35 | 70,000 | 7 |

**Number of Samples = 4**

### Python

`number_of_samples = X.shape[0]`

---

# 2. Number of Features

**Number of Features** = Number of input variables used to describe each sample.

The **target/output column is not counted as an input feature**.

| Age | Salary | Experience | Target |
| ---: | -----: | ---------: | -----: |
| 25 | 50,000 | 2 | 0 |
| 30 | 65,000 | 5 | 1 |
| 40 | 80,000 | 10 | 1 |

- Age → Feature
- Salary → Feature
- Experience → Feature
- Target → Output

**Number of Features = 3**

### Python

`number_of_features = X.shape[1]`

---

# 3. Sample-to-Feature Ratio

**Sample-to-Feature Ratio** = Number of samples available for each feature.

### Formula

$$
\text{Sample-to-Feature Ratio} =
\frac{\text{Number of Samples}}{\text{Number of Features}}
$$

### Example

Suppose a dataset has:

- Number of Samples = 1000
- Number of Features = 20

$$
\text{Sample-to-Feature Ratio} = \frac{1000}{20} = 50
$$

**Sample-to-Feature Ratio = 50**

This means:

**For every 1 feature, we have 50 samples.**

### Why It Matters

Consider two datasets:

| Dataset | Samples | Features | Sample-to-Feature Ratio |
| ------- | ------: | -------: | ----------------------: |
| A | 1000 | 10 | 100 |
| B | 1000 | 500 | 2 |

- Dataset A → 100 samples for every feature.
- Dataset B → 2 samples for every feature.

A **higher ratio** means we have more observations relative to the number of features.

A **low ratio** means the dataset has many features compared with the number of samples, which can make learning more difficult and may increase the risk of overfitting.

There is **no universal value** that defines a good Sample-to-Feature Ratio. It depends on the dataset and machine learning model.

### Python

`sample_feature_ratio = X.shape[0] / X.shape[1]`

# 4. Missingness Rate

**Missingness Rate** = Percentage of values in the dataset that are missing.

Missing values can appear as `NaN`, `None`, or `NULL`.

### Formula

$$
\text{Missingness Rate} =
\frac{\text{Number of Missing Values}}
{\text{Total Number of Values}}
\times 100
$$

### Example

| Age | Salary | Experience |
| ---: | -----: | ---------: |
| 25 | 50,000 | 2 |
| 30 | Missing | 5 |
| Missing | 80,000 | 10 |
| 35 | 70,000 | 7 |

- Number of Samples = 4
- Number of Features = 3
- Total Values = $4 \times 3 = 12$
- Missing Values = 2

$$
\text{Missingness Rate}
=
\frac{2}{12} \times 100
=
16.67\%
$$

**Missingness Rate = 16.67%**

This means **16.67% of the values in the dataset are missing**.

### Why It Matters

Missing data means some information about the samples is unavailable.

A dataset with a high missingness rate may require preprocessing such as:

- Removing rows or columns
- Filling missing values (Imputation)
- Using models that can handle missing values

Therefore, missingness can be used as an **intrinsic dataset quality characteristic**.

### Overall Missingness

Overall missingness measures missing values across the entire dataset.

`missingness_rate = (data.isnull().sum().sum() / data.size) * 100`

### Per-Feature Missingness

We can also calculate the missingness rate separately for each feature.

| Feature | Missingness Rate |
| ------- | ---------------: |
| Age | 25% |
| Salary | 25% |
| Experience | 0% |

`missingness_per_feature = data.isnull().mean() * 100`
