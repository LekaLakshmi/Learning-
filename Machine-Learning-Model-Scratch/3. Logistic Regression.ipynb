# Logistic Regression

- Logistic Regression is a supervised learning algorithm used for classification.
- First calculate the linear output: $z = Xw + b$
- Pass the linear output through the sigmoid function to get a probability between 0 and 1.

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

- For binary classification, we can use a threshold such as 0.5:
  - probability < 0.5 → class 0
  - probability >= 0.5 → class 1

### Weight Gradient

$$
dw = \frac{1}{m}X^T(\hat{y}-y)
$$

### Bias Gradient

$$
db = \frac{1}{m}\sum_{i=1}^{m}(\hat{y}_i-y_i)
$$

### Weight Update

$$
w = w - \alpha dw
$$

### Bias Update

$$
b = b - \alpha db
$$



```python
# import statements
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn import datasets

# Sigmoid function
def Sigmoid(x):
   return 1/( 1+np.exp(-x))

# Define class for logistic regression model 
class Logistic_Regression():
    def __init__(self,lr=0.00001,iteration=100000):
        self.lr=lr
        self.iteration=iteration
    def fit(self,x,y):
        sample,feature=x.shape
        self.weight=np.zeros(feature)
        self.bias=0
        for _ in range(self.iteration):
            y_hat= Sigmoid(np.dot(x,self.weight)+self.bias)
            weight_gradient=(np.dot(x.T,(y_hat-y)))/sample
            bias_gradient=((y_hat-y).sum())/sample
            self.weight-=self.lr*weight_gradient
            self.bias-=self.lr*bias_gradient
    def predict(self,x):
        y_hat= Sigmoid(np.dot(x,self.weight)+self.bias)
        prediction=[0 if i<0.5 else 1 for i in y_hat]
        return prediction
# Load the dataset
cancer_dataset=datasets.load_breast_cancer()
x,y=cancer_dataset.data,cancer_dataset.target
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=123)

# Make prediction
classification=Logistic_Regression()
classification.fit(x_train,y_train)
prediction=classification.predict(x_test)
print(f"Model prediction: {prediction}\n\n\nY_test: {y_test} ")


# Find accuracy
print(f"\n\n\nAccuracy of the model is {((prediction==y_test).sum())/len(y_test)} ")
```
