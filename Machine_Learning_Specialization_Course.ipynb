# Model Representation

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Housing prediction with two training sample
x_train=np.array([1,2]) # house size
y_train=np.array([300,500]) # house price
print(f"The features are {x_train}")
print(f"The targets are {y_train}")

# Get number of training sample
print(f"Shape of training set {x_train.shape}")
m=x_train.shape[0]
print(f"Total training samples are {m}")

# another way to get the total number of sample
m=len(x_train)
print(f"Total training samples are {m}")
```

```python
# Plot the training sample
plt.scatter(x_train,y_train,marker='o',c='r')
plt.title("Housing price prediction")
plt.xlabel("Size of the house")
plt.ylabel("Price of the house in 100K dollars")
plt.show()
```

### In Linear regression model , for different w and b values, we get differnt straight line

```python
# w=100 b=100
w=100
b=100

# Define function for model prediction
def compute_model_output(x,w,b):
  m=x.shape[0]
  f_wb=np.zeros(m)
  for i in range(m):
    f_wb[i]=x[i]*w+b
  return f_wb

prediction=compute_model_output(x_train,w,b,)
plt.plot(x_train,prediction,c='b',label='Prediction')
plt.scatter(x_train,y_train,c='r',marker='o',label='Actual Price')
plt.title("House Prediction")
plt.xlabel("House size")
plt.ylabel("House Price in 100K dollar")
plt.legend()
plt.show()
```

### Change the w and b and line also changes

```python
w=200
b=100
prediction=compute_model_output(x_train,w,b,)
plt.plot(x_train,prediction,c='b',label='Prediction')
plt.scatter(x_train,y_train,c='r',marker='o',label='Actual Price')
plt.title("House Prediction")
plt.xlabel("House size")
plt.ylabel("House Price in 100K dollar")
plt.legend()
plt.show()
```

### Predict price for this house size 1.2


```python
x_i=1.2
cost=x_i*w+b
print(f"Price of {x_i} size house is {cost} thousand dollars")
```

# Cost Function

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Use same training sample
x_train=np.array([1,2])
y_train=np.array([300,500])
```

```python
def compute_cost(x,y,w,b):
  m=x.shape[0]
  prediction=np.zeros(m)
  cost_sum=0
  for i in range(m):
    prediction[i]=x[i]*w+b
    cost=(prediction[i]-y[i])**2
    cost_sum=cost_sum+cost
  total_cost=(1/(2*m)) *cost_sum
  return total_cost
```
