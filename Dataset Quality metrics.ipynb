Phase 1: Dataset Structure 

* Number of Samples
* Number of Features
* Sample-to-Feature Ratio
* Missingness Rate
* Duplicate Sample Rate
* Matrix Sparsity


```python
import pandas as pd
```

```python
#Load the dataset
data=pd.read_csv('Dataset/Housing Price Prediction.csv')
```

## 1. Number of sample

```python
# Number of Samples
number_of_sample=data.shape[0]
print(f"The number of datapoints in the dataset: {number_of_sample} ")
```

## 2. Number of Features
 * Assumption: The last column is the target variable (Price), so it is excluded from the feature count. This results in 29 input features.

```python
# Number of features

number_of_feature=data.shape[1]-1
print(f"The Number of features in the dataset: {number_of_feature} ")
```

## 3.Sample Feature Ratio

$$
\text{Sample-to-Feature Ratio} =
\frac{\text{Number of Samples}}{\text{Number of Features}}
$$

```python
sample_feature_ratio=number_of_sample/number_of_feature
if sample_feature_ratio<5:
    print(f"Very Low sample feature ratio: {sample_feature_ratio:0.2f}")
elif sample_feature_ratio<10:
    print(f"Low sample feature ratio: {sample_feature_ratio:0.2f}")
elif sample_feature_ratio<20:
    print(f"Moderate sample feature ratio: {sample_feature_ratio:0.2f}")
else:
    print(f"Good sample feature ratio: {sample_feature_ratio:0.2f}")
```

## 4. Feature-wise Missingness Rate
$$
\text{Missingness Rate} =
\frac{\text{Total Number of Missing Values}}
{\text{Total Number of Data Cells}}
\times 100
$$

```python
missingness_rate_per_feature=((data.isnull().sum())/data.shape[0])*100
missingness_rate_per_feature
```

## 5. Duplicate Sample Rate


$$
\text{Duplicate Sample Rate} =
\frac{\text{Number of Duplicate Samples}}
{\text{Total Number of Samples}}
\times 100
$$

```python
duplicate_sample_rate=((data.duplicated().sum())/data.shape[0])*100
if duplicate_sample_rate ==0:
    print(f"No Duplicate Samples: {duplicate_sample_rate:.2f} ")
elif duplicate_sample_rate <5:
    print(f"Low Duplicate Sample Rate: {duplicate_sample_rate:.2f} ")
elif duplicate_sample_rate <20:
    print(f"Moderate Duplicate Sample Rate: {duplicate_sample_rate:.2f} ")
else:
    print(f"High Duplicate Sample Rate: {duplicate_sample_rate:.2f} ")
```

## 6. Matrix Sparsity

$$
\text{Matrix Sparsity} =
\frac{\text{Number of Zero Elements}}
{\text{Total Number of Elements}}
\times 100
$$

- If a large portion of the matrix is zero, we call it a **sparse matrix**.
- If most values are non-zero, it is called a **dense matrix**.

```python
Number_of_zeros=(data==0).sum().sum()
(Number_of_zeros/data.shape[0])*100
```
