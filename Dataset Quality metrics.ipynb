```python
import pandas as pd
```

```python
#Load the dataset
data=pd.read_csv('Dataset/Housing Price Prediction.csv')
```

## 1. Number of sample

```python
# Number of Samples
number_of_sample=data.shape[0]
print(f"The number of datapoints in the dataset: {number_of_sample} ")
```

## 2. Number of Features
 * Assumption: The last column is the target variable (Price), so it is excluded from the feature count. This results in 29 input features.

```python
# Number of features

number_of_feature=data.shape[1]-1
print(f"The Number of features in the dataset: {number_of_feature} ")
```

## 3.Sample Feature Ratio

$$
\text{Sample-to-Feature Ratio} =
\frac{\text{Number of Samples}}{\text{Number of Features}}
$$

```python
sample_feature_ratio=number_of_sample/number_of_feature
if sample_feature_ratio<5:
    print(f"Very Low sample feature ratio: {sample_feature_ratio:0.2f}")
elif sample_feature_ratio<10:
    print(f"Low sample feature ratio: {sample_feature_ratio:0.2f}")
elif sample_feature_ratio<20:
    print(f"Moderate sample feature ratio: {sample_feature_ratio:0.2f}")
else:
    print(f"Good sample feature ratio: {sample_feature_ratio:0.2f}")
```

## 4. Feature-wise Missingness Rate
$$
\text{Missingness Rate} =
\frac{\text{Total Number of Missing Values}}
{\text{Total Number of Data Cells}}
\times 100
$$

```python
missingness_rate_per_feature=((data.isnull().sum())/data.shape[0])*100
missingness_rate_per_feature
```

## 5. Duplicate Sample Rate


$$
\text{Duplicate Sample Rate} =
\frac{\text{Number of Duplicate Samples}}
{\text{Total Number of Samples}}
\times 100
$$

```python
duplicate_sample_rate=((data.duplicated().sum())/data.shape[0])*100
if duplicate_sample_rate ==0:
    print(f"No Duplicate Samples: {duplicate_sample_rate:.2f} ")
elif duplicate_sample_rate <5:
    print(f"Low Duplicate Sample Rate: {duplicate_sample_rate:.2f} ")
elif duplicate_sample_rate <20:
    print(f"Moderate Duplicate Sample Rate: {duplicate_sample_rate:.2f} ")
else:
    print(f"High Duplicate Sample Rate: {duplicate_sample_rate:.2f} ")
```

## 6. Coefficient of Variation

$$
\text{Coefficient of Variation (CV)} =
\frac{\text{Standard Deviation}}
{\text{Mean}}
\times 100
$$

- Standard Deviation: Measures the absolute spread of a numerical feature around its mean.
- Coefficient of Variation: Measures the relative spread of a numerical feature around its mean, enabling comparison of variability across features with different scales.

```python
print(f"Coefficient of Variation: \n{(data.std(numeric_only=True)/data.mean(numeric_only=True))*100} ")
```

# 7. Skewness

- Skewness measures the asymmetry of a dataset around its mean.
- Symmetric dataset: values are distributed almost equally on both sides of the mean.
- Left-skewed (Negative Skew): a few very small values and many larger values.
  - Example: 20, 50, 51, 52, 55, 56, 57
  - Long tail is on the left.
- Right-skewed (Positive Skew): a few very large values and many smaller values.
  - Example: 20, 21, 22, 23, 24, 25, 1000
  - Long tail is on the right.
- We cannot use the raw differences `(x - mean)` because their sum (or average) is always zero.
- Squaring the differences removes the sign, so it cannot distinguish left and right skew.
- Cubing the differences preserves the sign and gives more weight to values far from the mean.
- Skewness is unitless because it is normalized by the cube of the standard deviation.
- Interpretation:
  - Skewness < 0 → Left-skewed (Negative Skew)
  - Skewness > 0 → Right-skewed (Positive Skew)
  - Skewness ≈ 0 → Symmetric distribution

$$
\text{Skewness}
=
\frac{
\frac{1}{N}\sum_{i=1}^{N}(x_i-\mu)^3
}{
\sigma^3
}
$$

There are two common ways to calculate standard deviation:

1. Population Standard Deviation
   - Divide by N
   - ddof = 0

2. Sample Standard Deviation
   - Divide by N - 1
   - ddof = 1 (default in Pandas)

```python
numeric_data=data.select_dtypes(include="number")
(((((numeric_data-numeric_data.mean())**3).sum())/numeric_data.shape[0])/(numeric_data.std(ddof=0)**3))
```

# 8. Kurtosis

- Skewness: tells which side has the longer tail (left or right).
- Kurtosis: tells how many/how extreme the outliers (extreme values) are in the dataset.
- Kurtosis does not care whether the extreme values are on the left or right.
- We use the fourth power because we only care about how far the values are from the mean, not their direction.
- The fourth power removes the sign, so positive and negative outliers contribute equally.
- The fourth power gives much greater weight to extreme values than the square.
- Variance uses the square of the deviations to measure overall spread.
- Kurtosis uses the fourth power to make the metric much more sensitive to outliers.

Example:

Mean differences:
1   2   3   10

Square:
1   4   9   100

Fourth Power:
1   16   81   10000

$$
\text{Kurtosis}
=
\frac{
\frac{1}{N}\sum_{i=1}^{N}(x_i-\mu)^4
}{
\sigma^4
}
$$

```python
numeric_data=data.select_dtypes(include="number")
((((numeric_data-numeric_data.mean())**4).sum())/numeric_data.shape[0])/((numeric_data.std(ddof=0))**4)
```

# 9. Feature Scale Heterogeneity
is a single value computed for an entire dataset. It measures how similar or different the scales (magnitudes) of the numerical features are.

- Low Feature Scale Heterogeneity → Most features have similar scales (similar spread/magnitude).
- High Feature Scale Heterogeneity → Features have very different scales, meaning some features have much larger or smaller magnitudes than others.

$$
\text{Feature Scale Heterogeneity}
=
\frac{\operatorname{Std}\left(\sigma_1,\sigma_2,\ldots,\sigma_p\right)}
{\operatorname{Mean}\left(\sigma_1,\sigma_2,\ldots,\sigma_p\right)}
$$


```python
std_features=data.std(numeric_only=True,ddof=0)
print(f"Feature scale heterogeneity: {std_features.std(ddof=0)/std_features.mean()} ")
```

# 10. Pearson Correlation 
- measures the strength and direction of the linear relationship between two numerical features.
- Positive Correlation: When one feature increases, the other feature also tends to increase. Similarly, when one feature decreases, the other feature also tends to decrease.
- Negative Correlation: When one feature increases, the other feature tends to decrease, and vice versa.
