# Model Representation

we will use the example of housing price prediction. This lab will use a simple data set with only two data points 
| Size (1000 sqft)     | Price (1000s of dollars) |
| -------------------| ------------------------ |
| 1.0               | 300                      |
| 2.0               | 500                      |

```python

import numpy as np
import matplotlib.pyplot as plt

# Splitting the dataset into features and lables
train_features=np.array([1.0,2.0])
train_labels=np.array([300.0,500.0])
print(f"The training features are {train_features} and the training lables are {train_labels} ")
```

```python
# Get the size of the training set
m=train_features.shape[0]
print(f"The number of samples in the dataset are {m}")
m=len(train_features)
print(f"Another way to get number of samples in the dataset {m} ")
```

```python
# Plot the dataset 
plt.scatter(train_features,train_labels,marker='x',c='r')
plt.title("House Prediction")
plt.xlabel("Size (1000 sqft)")
plt.ylabel("Price (1000s of dollars)")
plt.show()
```

# Linear regression with one variable
Linear regression (which is a function that maps from `x` to `y`) is represented as 

$$ f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{1}$$

The formula above is how you can represent straight lines - different values of $w$ and $b$ give you different straight lines on the plot. <br/> <br/> <br/> <br/> <br/> 

Let's start with $w = 100$ and $b = 100$. 

```python


# Define function to make prediction
def model_output(x,w,b):
    m=x.shape[0]
    f_x=np.zeros(m)
    for i in range(m):
        f_x[i]=w*x[i]+b
    return f_x
```

```python
fig,ax=plt.subplots(1,4, figsize=(12,5))
# Initialize weight and bias
w=100
b=100

# Make prediction
prediction=model_output(train_features,w,b)
print(f" The prediction made by the model are {prediction} ")


# Plot the predicted value
ax[0].plot(train_features,prediction,c='b')
ax[0].scatter(train_features,train_labels,marker='x',c='r')

# Differnt weight and bias 
w=400
b=100
prediction=model_output(train_features,w,b)
print(f" The prediction made by the model are {prediction} ")
# Plot the predicted value
ax[1].plot(train_features,prediction,c='b')
ax[1].scatter(train_features,train_labels,marker='x',c='r')


# Differnt weight and bias 
w=230
b=10
prediction=model_output(train_features,w,b)
print(f" The prediction made by the model are {prediction} ")
# Plot the predicted value
ax[2].plot(train_features,prediction,c='b')
ax[2].scatter(train_features,train_labels,marker='x',c='r')


# Differnt weight and bias 
w=200
b=100
prediction=model_output(train_features,w,b)
print(f" The prediction made by the model are {prediction} ")
# Plot the predicted value
ax[3].plot(train_features,prediction,c='b')
ax[3].scatter(train_features,train_labels,marker='x',c='r')
plt.show()
```

# Prediction
Now that we have a model, we can use it to make our original prediction. Let's predict the price of a house with 1200 sqft. Since the units of $x$ are in 1000's of sqft, $x$ is 1.2.

```python
w=200
b=100
x=1.2
prediction=(w*x)+b
print(f" Model has predicted {prediction:.0f} thousand dollars for 1.2 square feet house size")
```
