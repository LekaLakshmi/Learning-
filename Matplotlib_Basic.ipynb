# Data visualization: 
* convert raw data into visuals
* Helps to see the patten easily
* In figure, we will have
    * x axis and y axis
    * x axis label and y axis label
    * Major tick mark 0,1,2,3,4,5
    * Minor tick mark 0.1,0.2,0.3,0.4...
    * Legend --> tells what each color line represent
    * Grid : helps to read the plot better
    * Title : title of the graph
    

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# create two variable and plot it

x=[1,2,3]
y=[4,5,6]
plt.plot(x,y)
plt.grid()
plt.show()
# (1,2) (2,5) (3,6) is plotted 
# with grid, it is easy to find points on the graph
```

# Univariate Category column

```python
#creating new column for the datafram with only catergories
df['Department']=['HR','Manager','Team Lead','new Joinee','Senior']*5
df.head()
```

# Pie Chart for Categories column

```python
#find out count of all categories

count=df['Department'].value_counts()
count
```

```python
plt.pie(count,labels=count.index,autopct='%1.1f',explode=[0,0,0,0,0.5]);
plt.axis("equal");

# count.index use this to get the label and %1.0 --> gives percentage without any decimal, for one decimal use %1.1 for two decimal use %1.2
# Use explode to make the pie to move from the center and to highlight --> change which category you want to explode and make others as 0, can also explode for 3 categories too
```

```python
plt.pie(count,labels=count.index,autopct='%1.1f',explode=[0,0,0.5,0,0.5]);
plt.axis("equal");
```

# Bar graph

```python
plt.bar(count.index,count,color=['blue','green','black','red','blue']);
plt.show()
```

# Bivariate ( two columns)

```python
# Create another numerical column

df['age']= [21, 23, 25, 22, 24,
       26, 27, 28, 29, 30,
       31, 32, 24, 26, 28,
       22, 21, 35, 33, 34,
       25, 27, 29, 30, 32]

df['salary']=[25000, 28000, 32000, 35000, 38000,
          42000, 46000, 50000, 54000, 58000,
          62000, 66000, 40000, 45000, 52000,
          34000, 30000, 85000, 78000, 82000,
          48000, 55000, 60000, 63000, 70000]
df.head()
```

# Scatter plot

```python
plt.scatter(df['age'],df['salary'],color='orange');
plt.show()
```

# Line plot

```python
plt.plot(df['age'],df['salary'],color='orange',linestyle='--',marker='*',linewidth='3')
plt.grid()
plt.show()
```

```python
# The output of the line graph is not proper as we have not sorted the age. 
sorted_age=df.sort_values("age")
plt.plot(sorted_age['age'],df['salary'],color='orange',linestyle='--',marker='*',linewidth='3')
plt.grid()
plt.show()
```

# Bar graph

# Bivariate 
* one numerical and one catergory columns

```python
# Filter the dataframe ane take only the hr salary
hr_sal=df[df['Department']=='HR']['salary']
hr_sal
```

```python
Manager_sal=df[df['Department']=='Manager']['salary']
Team_Lead_sal=df[df['Department']=='Team Lead']['salary']
new_Joinee_sal=df[df['Department']=='new Joinee']['salary']
Senior_sal=df[df['Department']=='Senior']['salary']
```

# Box plot
* each department min , max and median salary we can see
* also we can find if there is any outliers in each department

```python
plt.boxplot([hr_sal,
             Manager_sal,
             Team_Lead_sal,
             new_Joinee_sal,
             Senior_sal],

            labels=['HR',
                    'Manager',
                    'Team Lead',
                    'New Joinee',
                    'Senior']);

plt.show()
```

# Pie chart

```python
# find total salary in each department
salary_by_departement=df.groupby("Department")['salary'].sum()
salary_by_departement
```

```python
plt.pie(salary_by_departement,labels=salary_by_departement.index,autopct='%1.2f')
plt.axis('equal')
plt.show()
```

# Bar Plot

```python
# finding mean for each department
hr_mean=sum(hr_sal)/len(hr_sal)
Manager_mean=sum(Manager_sal)/len(Manager_sal)
Team_Lead_mean=sum(Team_Lead_sal)/len(Team_Lead_sal)
new_Joinee_mean=sum(new_Joinee_sal)/len(new_Joinee_sal)
Senior_mean=sum(Senior_sal)/len(Senior_sal)

plt.bar(['HR','Manager','team Lead','new Joinee','Senior'],[hr_mean,Manager_mean,Team_Lead_mean,new_Joinee_mean,Senior_mean],color=['red','green','blue','black','orange'])
```
