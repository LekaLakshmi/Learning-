# List Function

```python
lucky_number=[12,34,324,45,45,12,42,43,425,4525]
friends=["leka","Sona","Sruthy","Priya"]

# add two list
friends.extend(lucky_number)
print(friends)
friends=["leka","Sona","Sruthy","Priya"]

# append add at the end
friends.append("Jim")
print(friends)

# mention where you want to add the value
friends.insert(0,"Jackson")
print(friends)

# remove one elevemnt from list
friends.remove("leka")
print(friends)

# clear the list
friends.clear()
print(friends)

# pop removes last element
friends=["leka","leka","leka","leka","leka","Sona","Sruthy","Priya"]
friends.pop()
print(friends)

# find index of the element
print(friends.index("leka"))

# count number of entries
print(friends.count("leka"))

# sort in acending order
friends.sort()
print(friends)

# rever the list
friends.reverse()
print(friends)

# create a copy of the list
friends_new_list=friends.copy()
print(friends_new_list)
```

# Exponent Function

```python
def exponent(base_num,pow_num):
    result=1
    for index in range(pow_num):
        result=result*base_num
    return result

solution=exponent(10,2)
print(solution)
```

# Building translator which convert all vowels into letter g


```python

def translator(phrase):
    translated_word=""
    for letter in phrase:
        if letter.lower() in 'aeiou':
            if letter.isupper():
                translated_word=translated_word+'G'
            else:
                translated_word=translated_word+'g'            
        else:
            translated_word=translated_word+letter
    return translated_word

print(translator(input("enter the word to be translated")))
            
```

# Read file

* always good to close the file once it is opened

```python
employee_file=open("Employee.txt","r")

# below are the modes to open the file
# r read
# w modify the file
# a append at the end
# r+ read and write
print(employee_file.readable()) # to check if the file is readable
print(employee_file.read()) # display full file


employee_file.close()

employee_file=open("Employee.txt","r")
print(employee_file.readline()) # read first line
print(employee_file.readline()) # read second line
print(employee_file.readline()) # read third line

employee_file=open("Employee.txt","r")
print(employee_file.readlines()) # read all lines inside a list

employee_file=open("Employee.txt","r")
print(employee_file.readlines()[0]) # to access particular line


# to print all the lines
employee_file=open("Employee.txt","r")
for index in employee_file.readlines():
    print(index)
```
