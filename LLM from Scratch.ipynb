```python
with open("DOROTHY AND THE WIZARD IN OZ.txt",'r',encoding='utf-8') as f:
    text=f.read()
print(f"The length of the book is {len(text)}")
print(f"The first 500 character of the book is {text[:500]}")
character=sorted(set(text))
print(f"The unique characters in this book are {character} ")
print(f" The number of unique characters are {len(character)}  ")
```

# Notes
1. Tokenizer: Takes text and splits it into smaller units called tokens.
2. Encoder: Converts text (or tokens) into numbers.
3. Decoder: Converts numbers back into text.

# What I am doing here ?

1. Perform manual encoding and decoding.
2. Create a dictionary to map characters to numbers.
3. Create a dictionary to map numbers to characters.
4. Create an encoder function that converts text into token IDs.
5. Create a decoder function that converts token IDs back into text.

```python
Text_to_number= {char:i for i,char in enumerate(character)}
number_to_text={i:char for i, char in enumerate(character)}
encoder = lambda s: [Text_to_number[c] for c in s]
decoder= lambda l: ''.join([number_to_text[c] for c in l])
print(f" The encoded value : {encoder('hello')} ")
print(f" The decoded value is {decoder([60, 57, 64, 64, 67])} ")
```

# Notes
- Vocabulary Size = Number of unique tokens.
- Number of Tokens = Total tokens present in the text.

--------------------------------------------------------------------------------
**Character-level Tokenizer:** We split text into individual characters. Each character becomes one token. "hello" → ['h', 'e', 'l', 'l', 'o']

Advantages:
- Small vocabulary size (around tens or hundreds of unique characters).

Disadvantages:
- Long sequences (many tokens to encode and decode).
- More computation for the model.

------------------------------------------------------------------------------

**Word-level Tokenizer:**
We split text into words. Each word becomes one token. "I love cats" → ['I', 'love', 'cats']

Advantages:
- Shorter sequences (fewer tokens).

Disadvantages:
- Very large vocabulary.
- Cannot easily handle new or unseen words.

------------------------------------------------------------------------------
**Subword Tokenizer:**
A subword tokenizer is between a character-level tokenizer and a word-level tokenizer. "unbelievable" → ['un', 'believ', 'able']

Advantages:
- Smaller vocabulary than word-level tokenization.
- Shorter sequences than character-level tokenization.
- Can handle unseen words by breaking them into smaller pieces.

Modern LLMs such as GPT and BERT use subword tokenization.|
